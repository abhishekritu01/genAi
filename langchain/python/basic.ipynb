```python
# ==============================
# Python List - Notes
# ==============================

# Creating a list
my_list = [1, 2, 3, 4, 5]
print("Original list:", my_list)


# ==============================
# List Methods
# ==============================

# 1. append() → Adds element at the end
my_list.append(6)
print("After append(6):", my_list)  # Output: [1, 2, 3, 4, 5, 6]


# 2. insert(index, value) → Adds element at specific position
my_list.insert(0, 0)  # insert 0 at index 0
print("After insert(0, 0):", my_list)   # Output: [0, 1, 2, 3, 4, 5, 6]


# 3. remove(value) → Removes first occurrence of value
my_list.remove(3)
print("After remove(3):", my_list) # Output: [0, 1, 2, 4, 5, 6]


# 4. pop() → Removes last element (or index if given)
my_list.pop()
print("After pop():", my_list)  # Output: [0, 1, 2, 4, 5]

# pop with index
my_list.pop(0)
print("After pop(0):", my_list)  # Output: [1, 2, 4, 5] # removes element at index 0 (which is 0)


# 5. clear() → Removes all elements
temp_list = my_list.copy()
temp_list.clear()
print("After clear():", temp_list)  # Output: [] (original list remains unchanged)


# 6. index(value) → Returns index of first occurrence
print("Index of 4:", my_list.index(4))   # Output: 2 (index of first occurrence of 4)


# 7. count(value) → Counts occurrences of value
my_list.append(2)
print("Count of 2:", my_list.count(2))    # Output: 2 (there are two 2's in the list)


# 8. sort() → Sorts list in ascending order
my_list.sort()
print("After sort():", my_list)

# sort in descending order
my_list.sort(reverse=True)
print("After sort(reverse=True):", my_list)


# 9. reverse() → Reverses the list
my_list.reverse()
print("After reverse():", my_list)


# ==============================
# Extra Useful Operations
# ==============================

# Copy list
copy_list = my_list.copy()
print("Copied list:", copy_list)

# Length of list
print("Length:", len(my_list))

# Check if element exists
print("Is 2 in list?", 2 in my_list)

# Iterate list
print("Loop through list:")
for item in my_list:
    print(item)



# Slicing → list[start:stop:step]
my_list = [1, 2, 3, 4, 5]
print(my_list[1:4])    # [2, 3, 4]
print(my_list[:3])     # [1, 2, 3]
print(my_list[::2])    # [1, 3, 5]
print(my_list[::-1])   # [5, 4, 3, 2, 1]  ← reverse trick
```

```python

# ==============================
# TUPLE
# ==============================
# - Ordered, immutable (cannot change after creation)
# - Allows duplicates
# - Faster than list

my_tuple = (1, 2, 3, 4, 5)
print("Tuple:", my_tuple)

# Accessing elements (same as list)
print("First element:", my_tuple[0])       # 1
print("Slicing:", my_tuple[1:4])           # (2, 3, 4)

# Tuple methods (only 2!)
print("Count of 2:", my_tuple.count(2))    # 1
print("Index of 3:", my_tuple.index(3))    # 2

# Unpacking
a, b, c, d, e = my_tuple
print("Unpacked:", a, b, c, d, e)          # 1 2 3 4 5

# Single element tuple → needs trailing comma!
single = (42,)
print("Single tuple:", single, type(single))   # (42,) <class 'tuple'>
not_tuple = (42)
print("Not a tuple:", not_tuple, type(not_tuple))  # 42 <class 'int'>
```

```python
# ==============================
# SET
# ==============================
# - Unordered, no duplicates
# - Mutable (can add/remove)
# - No indexing (unordered)

my_set = {1, 2, 3, 4, 5, 2, 3}   # duplicates removed automatically
print("Set:", my_set)              # {1, 2, 3, 4, 5} (order not guaranteed)

# add() → adds one element
my_set.add(6)
print("After add(6):", my_set)

# remove() → raises error if not found
my_set.remove(6)
print("After remove(6):", my_set)

# discard() → no error if not found (safer)
my_set.discard(99)     # 99 doesn't exist, no error
print("After discard(99):", my_set)

# Set operations (the real power of sets)
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print("Union:", a | b)              # {1, 2, 3, 4, 5, 6}  → all elements
print("Intersection:", a & b)       # {3, 4}              → common elements
print("Difference:", a - b)         # {1, 2}              → in a but not b
print("Symmetric diff:", a ^ b)     # {1, 2, 5, 6}        → not in both
```

```python
# ==============================
# DICT
# ==============================
# - Key-value pairs
# - Ordered (Python 3.7+), mutable
# - Keys must be unique and immutable

my_dict = {"name": "Alice", "age": 25, "city": "Delhi"}
print("Dict:", my_dict)

# Accessing values
print("Name:", my_dict["name"])            # Alice
print("Age:", my_dict.get("age"))          # 25
print("Missing:", my_dict.get("x", "N/A")) # N/A (safe, no KeyError)

# Adding / Updating
my_dict["email"] = "alice@gmail.com"       # add new key
my_dict["age"] = 26                        # update existing key
print("Updated dict:", my_dict)

# Removing
my_dict.pop("city")
print("After pop('city'):", my_dict)

# Useful methods
print("Keys:", list(my_dict.keys()))       # ['name', 'age', 'email']
print("Values:", list(my_dict.values()))   # ['Alice', 26, 'alice@gmail.com']
print("Items:", list(my_dict.items()))     # [('name', 'Alice'), ...]

# Looping
for key, value in my_dict.items():
    print(f"{key} → {value}")
```

```python
# ==============================
# is  vs  ==
# ==============================


# ==  → checks VALUE equality    (are the values the same?)
# is  → checks IDENTITY equality (are they the SAME object in memory?)


# ==============================
# Basic example
# ==============================

a = [1, 2, 3]
b = [1, 2, 3]   # same values, but created separately

print(a == b)   # True  → same values
print(a is b)   # False → different obj
```

```python
# ==============================
# Shallow vs Deep Copy
# ==============================

import copy

# ==============================
# First — the 3 ways to "copy"
# ==============================

# 1. Assignment    → not a copy at all, just a new label on same object
# 2. Shallow copy  → new object, but inner objects are still shared
# 3. Deep copy     → completely independent copy, everything duplicated


# ==============================
# 1. Assignment (not a copy!)
# ==============================

a = [[1, 2], [3, 4]]
b = a                   # b is just another name for same object

b[0][0] = 99
print(a)    # [[99, 2], [3, 4]] ← a is affected!
print(b)    # [[99, 2], [3, 4]]

print(id(a) == id(b))   # True → same object


# ==============================
# 2. Shallow Copy
# ==============================

a = [[1, 2], [3, 4]]
b = a.copy()            # or copy.copy(a) or a[:]

print(id(a) == id(b))   # False → different outer objects ✅

b.append([5, 6])
print(a)    # [[1, 2], [3, 4]]      ← outer list safe ✅
print(b)    # [[1, 2], [3, 4], [5, 6]]

# BUT — inner objects are still shared!
b[0][0] = 99
print(a)    # [[99, 2], [3, 4]]     ← inner list affected! ❌
print(b)    # [[99, 2], [3, 4], [5, 6]]

print(id(a[0]) == id(b[0]))   # True → inner list is SAME object!


# ==============================
# 3. Deep Copy
# ==============================

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)    # completely independent copy

print(id(a) == id(b))         # False → different outer objects
print(id(a[0]) == id(b[0]))   # False → different inner objects too!

b[0][0] = 99
print(a)    # [[1, 2], [3, 4]] ← completely safe ✅
print(b)    # [[99, 2], [3, 4]]


# ==============================
# Visual to remember
# ==============================

#  Original:   a ──► [ [1,2] [3,4] ]
#                         ▲      ▲
#  Shallow:    b ──► [   │      │  ]   ← new outer list, same inner lists
#
#  Deep:       b ──► [ [1,2] [3,4] ]   ← everything brand new, no shared refs


# ==============================
# When does shallow copy work fine?
# ==============================

# If your list has only flat (non-nested) data → shallow copy is safe

a = [1, 2, 3, 4, 5]    # flat list, no nested objects
b = a.copy()

b[0] = 99
print(a)    # [1, 2, 3, 4, 5] ← safe! ints are immutable
print(b)    # [99, 2, 3, 4, 5]

# ints are immutable → modifying b[0] creates a NEW int object
# so a[0] is never touched


# ==============================
# Ways to shallow copy
# ==============================

a = [1, 2, 3]

b = a.copy()            # list method
b = list(a)             # list constructor
b = a[:]                # slicing
b = copy.copy(a)        # copy module

# all do the same thing for a flat list


# ==============================
# Ways to deep copy
# ==============================

b = copy.deepcopy(a)    # only one way — always use this for nested data

#Shallow copy creates a new outer object but still shares inner objects. Deep copy duplicates everything — fully independent#
```

```python
# ==============================
# Truthy vs Falsy
# ==============================


# In Python, every object has a boolean value
# You don't need to write == True or == False
# Python checks it automatically in if statements


# ==============================
# FALSY values → treated as False
# ==============================

# There are only 6 falsy values in Python — memorize these!

print(bool(0))          # False  ← zero int
print(bool(0.0))        # False  ← zero float
print(bool(""))         # False  ← empty string
print(bool([]))         # False  ← empty list
print(bool({}))         # False  ← empty dict
print(bool(()))         # False  ← empty tuple
print(bool(set()))      # False  ← empty set
print(bool(None))       # False  ← None


# ==============================
# TRUTHY values → everything else
# ==============================

print(bool(1))          # True
print(bool(-1))         # True   ← negative numbers too!
print(bool("hello"))    # True
print(bool([0]))        # True   ← list with one element, even if it's 0
print(bool(" "))        # True   ← space is not empty string!


# ==============================
# How it works in if statements
# ==============================

name = ""

if name:                        # same as if name != ""
    print("Name:", name)
else:
    print("Name is empty")      # this runs


my_list = []

if my_list:                     # same as if len(my_list) > 0
    print("List has items")
else:
    print("List is empty")      # this runs


# ==============================
# Common real world usage
# ==============================

# instead of this ❌
if len(my_list) == 0:
    print("empty")

# write this ✅
if not my_list:
    print("empty")

# instead of this ❌
if name != "" and name != None:
    print("valid name")

# write this ✅
if name:
    print("valid name")


# ==============================
# The tricky ones — interview favourites
# ==============================

print(bool(0))          # False ← obvious
print(bool("0"))        # True  ← "0" is a non-empty string!
print(bool([]))         # False ← obvious
print(bool([[]]))       # True  ← list containing an empty list → not empty!
print(bool(" "))        # True  ← space character → not empty string!
print(bool(-1))         # True  ← all non-zero numbers are truthy


# ==============================
# None vs False vs 0 vs "" → all falsy but not equal
# ==============================

print(None == False)    # False
print(None == 0)        # False
print(False == 0)       # True  ← interesting! bool is subclass of int
print(True == 1)        # True
print(True == 2)        # False


# ==============================
# Custom class — __bool__()
# ==============================

# Just like == has __eq__, truthiness has __bool__
# You can define truthy/falsy behaviour for your own class

class Cart:
    def __init__(self, items):
        self.items = items

    def __bool__(self):
        return len(self.items) > 0  # truthy if cart has items

cart1 = Cart([])
cart2 = Cart(["apple", "milk"])

print(bool(cart1))  # False → empty cart
print(bool(cart2))  # True  → has items

if cart2:
    print("Proceed to checkout")    # this runs
```
