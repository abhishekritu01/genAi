```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
from langchain_openai import ChatOpenAI

from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
model = ChatOpenAI()
```

```python
#create a state for the graph
class LLMState(TypedDict):
    question: str
    answer: str
```

```python
def llm_qa(state: LLMState) -> LLMState:
    
    #Extract the question from the state
    #from a prompt
    # ask the question to the llm
    # update the answer in the state and return the state
    
    question = state['question']
    prompt = f"Answer the following question: {question}"

    model_response = model.invoke(prompt).content

    state['answer'] = model_response

    return state
```

```python
graph = StateGraph(LLMState)

# create a node for the graph
graph.add_node('llm_qa',llm_qa)

# add edges to the graph
graph.add_edge(START, 'llm_qa')
graph.add_edge('llm_qa', END)


#compile the graph into a workflow
workflow = graph.compile()
```

```python
#execute the graph
initial_state ={ "question": "What is the capital of France?", "answer": "" }

final_state = workflow.invoke(initial_state)

# print(final_state)
print(final_state['answer'])
```
